This notebook runs multi-axial fatigue life predictions under cyclic high-temperature loading.

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl

mpl.rcParams['text.usetex'] = False
mpl.rcParams['mathtext.fontset'] = 'cm'

strains = np.linspace(0.001, 0.015, 50)
lives = []

for s in strains:
    N_f = 0.5 * ((s / 0.45) ** (1.0 / -0.6))
    lives.append(N_f)

plt.figure(figsize=(8, 5))
plt.loglog(strains, lives, 'b-', linewidth=2, label='FGM-SMA Continuum Life')
plt.xlabel(r'Equivalent Strain Amplitude ($\Delta\epsilon_{\mathrm{eq}}/2$)')
plt.ylabel(r'Reversals to Failure ($2N_f$)')
plt.title('Multi-Axial High-Cycle Thermal Fatigue Curve')
plt.grid(True, which='both', ls='--')
plt.legend()
plt.show()
```
